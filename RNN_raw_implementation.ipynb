```python
import torch
import torch.nn as nn
```

```python
class RNNcustom(nn.Module):
  def __init__(self):
    super().__init__()
    self.rnn = nn.RNN(input_size=5,hidden_size=3,batch_first=True)
    self.fully_connected = nn.Linear(3,1)
    self.sigmoid = nn.Sigmoid()


  #forward pass
  def forward(self,x):
    out,_ = self.rnn(x)
    out = out[:-1:]
    out = self.fully_connected(out)
    out = self.sigmoid(out)
    return out
```

```python
model = RNNcustom()
model
```
